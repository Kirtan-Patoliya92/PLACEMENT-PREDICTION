```python
categorical_datas = ['gender','city_tier','ssc_board','hsc_board','hsc_stream', 'internships_count', 'projects_count',
       'certifications_count','leadership_roles','extracurricular_activities','backlogs']
from scipy.stats import chi2_contingency

chi2_result = {}
for col in categorical_datas:
    contingency = pd.crosstab(df[col],df['placed'])
    chi2_stats,p_val,dof,exfq = chi2_contingency(contingency)
    decision = 'Keep Feature' if p_val < 0.05 else "Drop Feature"

    chi2_result[col] = {
        'chi2_statistics' : chi2_stats,
        'p_value' : p_val,
        'Decision' : decision
    }

chi2_df = pd.DataFrame(chi2_result).T
chi2_df = chi2_df.sort_values(by = 'p_value')
chi2_df
```

```python
fianl_data = df[['age','ssc_percentage',
       'hsc_percentage', 'hsc_stream', 'degree_percentage',
       'degree_field', 'mba_percentage', 'internships_count', 'projects_count',
       'certifications_count', 'technical_skills_score', 'soft_skills_score',
       'aptitude_score', 'communication_score', 'work_experience_months',
       'leadership_roles', 'extracurricular_activities', 'backlogs', 'placed',
       ]]
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
x = fianl_data.drop('placed',axis=1)
y = fianl_data['placed']
scaler = StandardScaler()
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.20,random_state=42)
x_train = scaler.fit_transform(x_train)
x_test = scaler.fit_transform(x_test)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.tree import DecisionTreeClassifier
from sklearn.svm import SVC
from sklearn.ensemble import RandomForestClassifier
from sklearn.ensemble import AdaBoostClassifier,GradientBoostingClassifier
from xgboost import XGBClassifier
```

```python
models = {
    'Logistic' : LogisticRegression(random_state=42),
    'NB' : GaussianNB(),
    'Trees' : DecisionTreeClassifier(),
    'KNN' : KNeighborsClassifier(),
    'SVM' : SVC(),
    'RendomF' : RandomForestClassifier(
        n_estimators=100,
        max_depth=None,
        random_state=42
    ),
    'Ada_Boost' : AdaBoostClassifier(
        n_estimators=100,
        random_state=42
    ),
    'Gradiant_Boost' : GradientBoostingClassifier(
        n_estimators=100,
        learning_rate=0.1,
        random_state=42
    ),
    'XGBoost' : XGBClassifier(
        n_estimators = 100,
        learning_rate = 0.1,
        max_depth = 3,
        use_label_encoder = False,
        eval_metric = 'mlogloss',
        random_state = 42
    )
}
```

```python
data = []
from sklearn.metrics import accuracy_score,f1_score
for model,predict in models.items():
    predict.fit(x_train,y_train)
    y_pred = predict.predict(x_test)
    acc = accuracy_score(y_test,y_pred)
    f1 = f1_score(y_test,y_pred)

    data.append(
        {
            'Model_name' : model,
            'Accuracy' : acc,
            'f1_score' : f1
        }
    )
```

```python
for i in data:
    print(i)
```
